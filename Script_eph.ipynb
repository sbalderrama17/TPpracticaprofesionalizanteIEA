```python
import pandas as pd
from google.colab import drive

#carga del dataframe desde el archivo .csv
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/eph_tecnologia_microdatos.csv')

#calculo de metricas de manera individual sobre el ingreso total familiar
media = df['ITF'].mean()
mediana = df['ITF'].median()
minimo = df['ITF'].min()
maximo = df['ITF'].max()
desviacion_estandar = df['ITF'].std()

#resumen completo con .describe()
resumen_completo = df['ITF'].describe()

# 5. Muestra de todos los datos obtenidos
print("CÁLCULOS INDIVIDUALES SOBRE ITF")
print("Media:", media)
print("Mediana:", mediana)
print("Mínimo:", minimo)
print("Máximo:", maximo)
print("Desviación Estándar:", desviacion_estandar)

print("----------------")
print("RESUMEN COMPLETO")
print(resumen_completo)
```
